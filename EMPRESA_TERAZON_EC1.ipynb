```python
# La variable y las listas
meses = ["2023-01", "2023-02", "2023-03", "2023-04", "2023-05", "2023-06",
         "2023-07", "2023-08", "2023-10", "2023-11", "2023-12", "2024-01"]
inversion_publicidad = [2060, 4270, 3440, 2980, 2900, 4990, 5540, 2050, 2630, 5840, 5620, 4250]
ventas = [21409, 36413, 29059, 27023, 29338, 40407, 41164, 22082, 30668, 43695, 42991, 40514]

total_inversion = 0
total_ventas = 0
cantidad_alta = 0
cantidad_media = 0
cantidad_baja = 0
mejor_mes = ""
mejor_ratio = 0

# Bucle for
print("Mes     | Inversión | Ventas  | Ventas por sol | Efectividad")
print("-" * 62)

for i in range(len(meses)):
    ratio = ventas[i] / inversion_publicidad[i]

    if ratio >= 10:
        nivel = "Alta"
        cantidad_alta += 1
    elif ratio >= 8:
        nivel = "Media"
        cantidad_media += 1
    else:
        nivel = "Baja"
        cantidad_baja += 1

    total_inversion += inversion_publicidad[i]
    total_ventas += ventas[i]

    if ratio > mejor_ratio:
        mejor_ratio = ratio
        mejor_mes = meses[i]

    print(f"{meses[i]} | {inversion_publicidad[i]:>9} | {ventas[i]:>7} | {ratio:>14.2f} | {nivel}")

# Promedios
promedio_inversion = total_inversion / len(meses)
promedio_ventas = total_ventas / len(meses)
ventas_por_sol = total_ventas / total_inversion

print()
print("RESUMEN")
print(f"Inversión promedio mensual: S/ {promedio_inversion:,.2f}")
print(f"Ventas promedio mensuales : S/ {promedio_ventas:,.2f}")
print(f"Ventas por cada sol invertido (promedio): S/ {ventas_por_sol:.2f}")
print(f"Meses con efectividad Alta: {cantidad_alta} | Media: {cantidad_media} | Baja: {cantidad_baja}")
print(f"Mejor mes: {mejor_mes} (S/ {mejor_ratio:.2f} de ventas por sol)")

# Bucle while
print()
inversion_plan = 0
while inversion_plan <= 0:
    texto = input("Ingrese la inversión publicitaria planificada (S/): ")
    try:
        inversion_plan = float(texto)
    except ValueError:
        inversion_plan = 0
    if inversion_plan <= 0:
        print("Dato no válido. Ingrese un número mayor a 0.")

# ventas estimadas
ventas_estimadas = inversion_plan * ventas_por_sol
print(f"Con una inversión de S/ {inversion_plan:,.2f}, las ventas estimadas serían de S/ {ventas_estimadas:,.2f}")

# fin
if inversion_plan > promedio_inversion:
    print("La inversión planificada está POR ENCIMA del promedio histórico.")
else:
    print("La inversión planificada está en o por DEBAJO del promedio histórico.")
```
