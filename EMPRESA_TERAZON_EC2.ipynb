ADMINISTRACIÓN DE LA TABLA DE DATOS DE LA EMPRESA


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from IPython.display import display

#todo lo de arriba son full importaciones para el funcionamiento de nuestro programa :v

ARCHIVO = "datos_empresa.xlsx"
COL_X = "inversion_publicidad"
COL_Y = "ventas"

try:
  if ARCHIVO.endswith(".xlsx"):
    df = pd.read_excel(ARCHIVO, sheet_name=0)
  else:
    df = pd.read_csv(ARCHIVO)
  print("se cargo el archivo con exito:", ARCHIVO)
except FileNotFoundError:
    print("Lo sentimos pero no se encontró el archivo")
df.head()
```

INSPECCION INICIAL DE LOS DATOS

```python
print("Dimensiones (filas, columnas):",df.shape)
print()
df.info()
print()
print("Celdas nulas por cada columna:")
print(df.isnull().sum())
print()
print("Filas que se duplican:", df.duplicated().sum())

df.describe()
```

LIMPIEZA Y PREPARACIÓN




```python
filas_antes = len(df)

df = df.drop_duplicates()

df = df.dropna(subset=[COL_Y])

filas_despues = len(df)

print(f"Filas iniciales: {filas_antes}")
print(f"Filas después de la limpieza: {filas_despues}")
print(f"Filas eliminadas: {filas_antes - filas_despues}")
print("\nCeldas nulas actuales:")
print(df.isnull().sum())
```

ANÁLISIS EXPLORATORIO (EDA)

```python
print("Estadísticas descriptivas (datos limpios):")

fig, ax = plt.subplots(2, 3, figsize=(16, 8))

ax[0, 0].scatter(df[COL_X], df[COL_Y], alpha=0.6)
ax[0, 0].set_title("Relación entre X e Y")
ax[0, 0].set_xlabel(COL_X); ax[0, 0].set_ylabel(COL_Y)

ax[0, 1].hist(df[COL_X], bins=12)
ax[0, 1].set_title("Distribución de " + COL_X)

ax[0, 2].hist(df[COL_Y], bins=12)
ax[0, 2].set_title("Distribución de " + COL_Y)

ax[1, 0].boxplot(df[COL_X]); ax[1, 0].set_title("Boxplot de " + COL_X)
ax[1, 1].boxplot(df[COL_Y]); ax[1, 1].set_title("Boxplot de " + COL_Y)

if "mes" in df.columns:
    ax[1, 2].plot(df["mes"], df[COL_Y], marker="o")
    ax[1, 2].set_xticks(range(0, len(df), 3))
    ax[1, 2].tick_params(axis="x", rotation=90)
    ax[1, 2].set_title(COL_Y + " a lo largo del tiempo")
else:
    ax[1, 2].axis("off")

plt.tight_layout(); plt.show()

print("Correlación entre X e Y:", round(df[COL_X].corr(df[COL_Y]), 3))
```
